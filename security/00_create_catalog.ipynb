```python
%python
dbutils.widgets.text("storage_acount", ".", "Adls")
storage_acount = dbutils.widgets.get("storage_acount")
dbutils.widgets.text("container", ".", "container")
container = dbutils.widgets.get("container")
```

```python
%python
abfs = f"abfss://{container}@{storage_acount}.dfs.core.windows.net/"
```

```python
spark.sql(f"""
CREATE CATALOG IF NOT EXISTS catalog_dev 
MANAGED LOCATION '{abfs}'
""")
```

```python
spark.sql(f"""
CREATE CATALOG IF NOT EXISTS catalog_qa 
MANAGED LOCATION '{abfs}'
""")
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS catalog_dev.bronze;
CREATE SCHEMA IF NOT EXISTS catalog_dev.silver;
CREATE SCHEMA IF NOT EXISTS catalog_dev.gold;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS catalog_dev.bronze
MANAGED LOCATION 'abfss://{container}@{storage_acount}.dfs.core.windows.net/bronze/';
```

```python
%sql
CREATE TABLE IF NOT EXISTS catalog_dev.bronze.customers (
    customer_id INT,
    first_name STRING,
    last_name STRING,
    date_of_birth DATE,
    email STRING,
    phone_number STRING,
    created_at TIMESTAMP
);

CREATE TABLE IF NOT EXISTS catalog_dev.bronze.loans (
    loan_id INT,
    customer_id INT,
    loan_amount DECIMAL(18,2),
    interest_rate DECIMAL(5,2),
    start_date DATE,
    end_date DATE,
    status STRING
);

CREATE TABLE IF NOT EXISTS catalog_dev.bronze.payments (
    payment_id STRING,
    loan_id INT,
    payment_date DATE,
    payment_amount DECIMAL(18,2),
    payment_status STRING
);
```
